Step  1 - Memasukkan Library

```python
import numpy as np
import pandas as pd
import cv2 as cv
from google.colab.patches import cv2_imshow # untuk display image
from skimage import io
from skimage import transform
from PIL import Image
import matplotlib.pylab as plt
```

Step 2 - Menampilkan Image

```python
# list url setiap gambar
urls = ["https://iiif.lib.ncsu.edu/iiif/0052574/full/800,/0/default.jpg",
        "https://iiif.lib.ncsu.edu/iiif/0016007/full/800,/0/default.jpg",
        "https://i.pinimg.com/736x/3a/d3/9b/3ad39bd0abb4d280d7c65fb4bd9277b5.jpg"]
# loop untuk setiap gambar
for url in urls:
  image = io.imread(url) # membaca url gambar
  image = cv.resize(image, (0,0), fx=0.5, fy=0.5) # mengatur size gambar
  image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB) # convert gambar dari BGR ke RGB
  final_frame = cv.hconcat((image, image_2)) # menggabungkan kedua gambar
  cv2_imshow(final_frame) # menampilkan hasil gambar
  print('\n')
```

Step 3 - Membaca Resolusi Gambar

```python
tinggi = image_2.shape[0]
lebar = image_2.shape[1]
print("resolusi image: tinggi x lebar = ", tinggi , " x ", lebar)
cv2_imshow(image_2)
```

Step 4 - Membuat Garis Pada Gambar

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

# buat garis horizontal ditengah gambar
for y in range (lebar):
  image_3[int((tinggi)/2),y] = [255,255,255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

Tugas Praktikum

1. Membuat Garis Vertikal dan Garis Menyilang Diagonal

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

# Garis Vertikal
for x in range (tinggi):
  image_3[x, int((lebar)/2)] = [255,255,255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

for y in range(tinggi):

    # 1. Diagonal Kiri Atas ke Kanan Bawah
    x1 = int(y * (lebar / tinggi))
    image_3[y, x1] = [255, 255, 255]

    # 2. Diagonal Kanan Atas ke Kiri Bawah
    x2 = lebar - 1 - x1
    image_3[y, x2] = [255, 255, 255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

2. Membuat Garis Horizontal di Tengah Gambar

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

# Buat garis horizontal di tengah gambar
titik_awal = 100
titik_akhir = 200

for x in range(titik_awal, titik_akhir):
  image_3[int((tinggi)/2), x] = [255, 255, 255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

3. Membuat Kotak Putih di Sembarang Tempat

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

# Buat gambar kotak di sembarang tempat
for y in range(50, 150):
  for x in range(200, 300):
    image_3[y, x] = [255, 255, 255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```
